```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.io.wavfile as wav
from tkinter import Tk, filedialog, Button, Label, Frame
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg
import os

# np.diff implementa la Primera Diferencia hacia Adelante (Forward Difference):
#   Δx[n] = x[n+1] - x[n]
# Es la aproximación discreta de la primera derivada de la señal.

def cargar_y_graficar():
    ruta_archivo = filedialog.askopenfilename(
        filetypes=[("Archivos WAV", "*.wav")],
        title="Seleccione el archivo de audio"
    )

    if not ruta_archivo:
        return

    try:
        frecuencia_muestreo, datos = wav.read(ruta_archivo)

        if len(datos.shape) == 2:
            datos = datos[:, 0]

        # Extraer primeras 30 muestras y calcular Forward Difference (Primera Derivada Discreta)
        datos_30 = datos[:30]
        diferencia_hacia_adelante = np.diff(datos_30)  # Δx[n] = x[n+1] - x[n], 29 valores

        tiempo_delta = np.linspace(0, 29 / frecuencia_muestreo, num=29)

        for widget in frame_grafica.winfo_children():
            widget.destroy()

        fig, ax = plt.subplots(figsize=(8, 6))
        ax.stem(tiempo_delta, diferencia_hacia_adelante, basefmt=" ", linefmt='r-', markerfmt='ro')
        ax.set_title('Primera Derivada Discreta (Forward Difference) — Primeras 29 Muestras', fontsize=13)
        ax.set_xlabel('Tiempo [s]', fontsize=12)
        ax.set_ylabel('Amplitud Δx[n]', fontsize=12)
        ax.grid(True)

        canvas = FigureCanvasTkAgg(fig, master=frame_grafica)
        canvas.draw()
        canvas.get_tk_widget().pack()

    except Exception as e:
        print(f"Error al cargar el archivo: {e}")

ventana = Tk()
ventana.title("Visualizador de Primera Derivada Discreta (WAV)")
ventana.geometry("800x600")

Label(ventana, text="Seleccione un archivo WAV para graficar su Forward Difference", font=("Arial", 12)).pack(pady=10)
Button(ventana, text="Cargar Archivo de Audio", command=cargar_y_graficar, font=("Arial", 12)).pack(pady=10)

frame_grafica = Frame(ventana)
frame_grafica.pack(pady=20, expand=True, fill='both')

ventana.mainloop()
```
