```python
import tkinter as tk
import numpy as np

# Nota: se usa implementación manual en lugar de numpy.convolve para mostrar
# explícitamente el mecanismo de cada método (educativo). numpy.convolve
# usa FFT internamente para secuencias largas y es más eficiente en producción.

def convolucion_directa(X, h):
    """Convolución por método directo: deslizamiento de h sobre X invertida."""
    M = len(X)
    N = len(h)
    P = M + N - 1
    Y_directo = np.zeros(P)

    for n in range(P):
        # Para cada salida n, acumular h[i]*X[n-i] para todo i en h
        X_values = [X[n-i] if (n-i) >= 0 and (n-i) < len(X) else 0 for i in range(N)]
        Y_directo[n] = sum(h[i] * X_values[i] for i in range(N))

    return Y_directo

def convolucion_suma(X, h):
    """Convolución por suma directa: Y[n] = sum_k X[k]*h[n-k]."""
    M = len(X)
    N = len(h)
    P = M + N - 1
    Y_suma = np.zeros(P)

    for n in range(P):
        for k in range(N):
            if 0 <= n - k < M:
                Y_suma[n] += X[n - k] * h[k]

    return Y_suma

def calcular_convolucion():
    try:
        X = list(map(float, entry_X.get().split(',')))
        h = list(map(float, entry_h.get().split(',')))

        Y_directo = convolucion_directa(X, h)
        Y_suma = convolucion_suma(X, h)

        label_result_directo.config(text=f"Método Directo: {[round(v,4) for v in Y_directo.tolist()]}")
        label_result_suma.config(text=f"Método Suma:    {[round(v,4) for v in Y_suma.tolist()]}")

    except ValueError:
        label_result_directo.config(text="Error: ingresa números separados por comas")
        label_result_suma.config(text="")

root = tk.Tk()
root.title("Calculadora de Convolución Discreta")

tk.Label(root, text="Secuencia X[n] (separada por comas):").pack(pady=5)
entry_X = tk.Entry(root, width=50)
entry_X.pack(pady=5)

tk.Label(root, text="Secuencia h[n] (separada por comas):").pack(pady=5)
entry_h = tk.Entry(root, width=50)
entry_h.pack(pady=5)

tk.Button(root, text="Calcular Convolución", command=calcular_convolucion).pack(pady=20)

label_result_directo = tk.Label(root, text="", fg="blue", wraplength=450)
label_result_directo.pack(pady=5)
label_result_suma = tk.Label(root, text="", fg="green", wraplength=450)
label_result_suma.pack(pady=5)

root.mainloop()
```
